```python
import pandas as pd

# 读取CSV文件并仅选择第2、3、4列
df = pd.read_csv('dataset.csv', usecols=[1, 2, 3])
# 清洗除M T D三类传感器以外的数据
df = df[df['did'].str[0].isin(['M', 'T', 'D'])]

# 打印提取的子集
print(df)
```

```python
# 规则字典
rule_dict = {'M': 'motion', 'T': 'temperature', 'D': 'door'}

# 根据规则生成func列
func = df['did'].str[0].map(rule_dict)

df.insert(2, 'func', func)
# df=df.head(1000)
```

```python
df.to_csv('processed_dataset.csv', index=False)
```
