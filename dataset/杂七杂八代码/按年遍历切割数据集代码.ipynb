```python
import pandas as pd  
import os 
import glob 
```

```python
# 源文件夹路径  
src_folder = 'D:\\python\\ly\\处理好的hh系列数据集\\上传至微盘文件' 
```

```python
# 遍历源文件夹下所有的hh101~hh130文件夹  
for folder_name in range(101, 121): 
    save_path = os.path.join(src_folder, f'hh{folder_name}')
    folder_path = os.path.join(src_folder, f'hh{folder_name}\\data.csv')
    # 读取 data.csv 文件  
    df = pd.read_csv(folder_path)  
    # 将 ts 列转换为 datetime 类型  
    df['ts'] = pd.to_datetime(df['ts'])  
    # 获取当前年份  
    current_year = int(df['ts'].iloc[0].year)  
    # 循环切割并保存每个年份的数据  
    for i in range(len(df)):  
        # 获取当前行的年份  
        year = int(df['ts'].iloc[i].year)  
      
        # 如果当前年份与之前年份不同，则新建一个 csv 文件并保存数据  
        if year != current_year:  
            # 将之前的年份数据保存为 csv 文件  
            df[df['ts'].dt.year == current_year].to_csv(os.path.join(save_path, f'data_{current_year}.csv'), index=False)  
          
            # 更新当前年份为新的年份  
            current_year = year  
          
    # 将最后一个年份的数据保存为 csv 文件  
    df[df['ts'].dt.year == current_year].to_csv(os.path.join(save_path, f'data_{current_year}.csv'), index=False)
```
