```python
import os
import shutil
```

```python
# 源文件夹路径  
src_folder = 'D:\\python\\ly\\处理好的hh系列数据集' 
# 目标文件夹路径  
dst_folder = 'D:\\python\\ly\\处理好的hh系列数据集\\上传至微盘文件' 
```

```python
# 遍历源文件夹下所有的hh101~hh130文件夹  
for folder_name in range(101, 131):  
    folder_path = os.path.join(src_folder, f'hh{folder_name}')  
    if os.path.exists(folder_path):  
        # 进入每个文件夹，查找data.csv, env.csv, readme-ly.txt文件  
        for file_name in ['data.csv', 'env.csv', 'readme-ly.txt']:  
            file_path = os.path.join(folder_path, file_name)  
            if os.path.exists(file_path):  
                # 在目标文件夹下创建对应名称的文件夹  
                dst_folder_name = f'hh{folder_name}'  
                dst_folder_path = os.path.join(dst_folder, dst_folder_name)  
                if not os.path.exists(dst_folder_path):  
                    os.makedirs(dst_folder_path)  
                # 复制文件到目标文件夹  
                shutil.copy(file_path, dst_folder_path)
```
